```python
#Water jug problem using Deapth First Search(DFS)
#------------------------------------------------

def water_jug_dfs(x, y, target):
  visited = set()
  path = []

  def dfs(a, b):
    #Base Case: Goal Reached
    if a == target or b == target:
      path.append((a, b))
      print("Goal Reached")
      for step in path:
        print(step)
      return True

   #Mark current state as visited
    if (a, b) in visited:
      return False
    visited.add((a, b))
    path.append((a, b))

    #All possible next moves(6 steps)
    moves = [
        (x, b),
        (a, y),
        (0, b),
        (a, 0),

        (a - min(a, y - b), b + min(a, y - b)),
        (a + min(b, x - a), b - min(b, x - a))
        ]

    #Try each next state (DFS Recursion)
    for new_a, new_b in moves:
      if (new_a, new_b) not in visited:
        if dfs(new_a, new_b):
          return True

    #Backtrack if not found
    path.pop()
    return False

  #Start DFS from (0, 0)
  if not dfs(0, 0):
    print("No Solution Found")

#Example Run
x = 4
y = 3
target = 2

water_jug_dfs(x, y, target)
```
