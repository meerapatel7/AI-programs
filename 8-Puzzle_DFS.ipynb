```python
# Function to print the puzzle nicely
def print_puzzle(state):
  for i in range(0, 9, 3):
    print(state[i:i+3])
  print()

# Function to get possible moves
def get_neighbours(state):
  neighbours = []
  index = state.index('0')

  moves = {
      0 : [1, 3],
      1 : [0, 2, 4],
      2 : [1,5],
      3 : [0, 4, 6],
      4 : [1, 3, 5, 7],
      5 : [2, 4, 8],
      6 : [3, 7],
      7 : [4, 6, 8],
      8 : [5, 7],
  }

  for move in moves[index]:
    new_state = list(state)
    new_state[index], new_state[move] = new_state[move], new_state[index]
    neighbours.append(''.join(new_state))

  return neighbours

#DFS recursive function
def dfs(state, goal, visited, path, max_depth):
  if state == goal:
    return path

  if max_depth == 0:
    return None

  visited.add(state)

  for neighbour in get_neighbours(state):
    if neighbour not in visited:
      # Create a copy of visited set for each recursive call to avoid modifying shared set
      result = dfs(neighbour, goal, visited.copy(), path + [neighbour], max_depth - 1)
      if result:
        return result

  return None

#Driver Code
start = '125340678'
goal = '123456780'

initial_path = [start]
visited_states = set()
max_search_depth = 40 # You can adjust this value

solution_path = dfs(start, goal, visited_states, initial_path, max_search_depth)

if solution_path:
  print("Goal Reached!")
  print("Number of moves:", len(solution_path) - 1)
  for step in solution_path:
    print_puzzle(step)
else:
  print(f"No solution found within depth limit of {max_search_depth}")
```
